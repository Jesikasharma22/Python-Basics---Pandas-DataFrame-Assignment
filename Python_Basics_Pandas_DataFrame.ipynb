#2. Write a Python program to count how many times a number appears in a list. Example List [2,3,4,2,5,2]

Solution:
```
numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter the number to count: "))

count = numbers.count(num)

print("The number", num, "appears", count, "times.")
```

#4. Create a new column called Revenue, where: Revenue = Quantity × Price Display the updated dataset with the new column.

#7. Find all records where: Region = "South" and Quantity > 2

Solution:
```
south_data = df[(df["Region"] == "South") & (df["Quantity"] > 2)]

print(south_data)
```

To display selected columns:
```
south_data = df[
    (df["Region"] == "South") &
    (df["Quantity"] > 2)
]

print(south_data[
    ["Transaction ID", "Product Category", "Quantity", "Price", "Region"]
])
```


#9. Create a random array of 10 numbers and: 1.Find max and min 2.Sort the array

Solution:
```
import numpy as np

# Generate random array of 10 numbers
arr = np.random.randint(1, 100, 10)

print("Original Array:")
print(arr)

# Maximum
print("Maximum:", np.max(arr))

# Minimum
print("Minimum:", np.min(arr))

# Sorted array
print("Sorted Array:")
print(np.sort(arr))
```

#10. Replace all even numbers with 0: arr = np.array([1,2,3,4,5,6])

Solution :

```
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

arr[arr % 2 == 0] = 0

print(arr)
```
